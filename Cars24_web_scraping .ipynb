```python
url="https://www.cars24.com/buy-used-cars-delhi-ncr/?sort=bestmatch&serveWarrantyCount=true&storeCityId=1"
```

```python
res=requests.get(url)
print(res)
```

```python
data=BeautifulSoup(res.text,'lxml')
data
```

#### Extract car manf Yr,Name,Brand

```python
# <span id="" class="sc-bcXHqh bAcffq">2019 Renault Kwid</span>
```

```python
n1 = data.find('span', attrs={'class':"sc-bcXHqh bAcffq"})
print(n1)
print(n1.text)
```

```python
cnames = []
for i in data.find_all('span', attrs={'class':"sc-bcXHqh bAcffq"}):
    cnames.append(i.text)
```

```python
len(cnames)
```

```python
#<span id="" class="sc-bcXHqh bKVBht">&nbsp;RXT 1.0 (O)</span>
```

```python
models=[]
for i in data.find_all("span",attrs={'class':"sc-bcXHqh bKVBht"}):
    models.append(i.text)
```

```python
len(models)
```

```python
models
```

```python
#<p id="" class="sc-bcXHqh kNDBvu">42,793 km</p>
```

```python
for i,j in enumerate(data.find_all('p',attrs={'class':'sc-bcXHqh kNDBvu'})):
    print(i,j.text)
```

```python
dist,fuel=[],[]
transm,regis=[],[]
for i,j in enumerate(data.find_all('p',attrs={'class':'sc-bcXHqh kNDBvu'})):
    if i%4==0:
        dist.append(j.text)
    elif i%4==1:
        fuel.append(j.text)
    elif i%4==2:
        transm.append(j.text)
    elif i%4==3:
        regis.append(j.text)
```

```python
print(len(dist),len(fuel),len(transm),len(regis))
```

```python
# <p>EMI ₹3,584/m*</p>
#<div class="styles_flexItem__2z__J"><span class="sc-dPWrhh gGOxIv"><p>EMI ₹3,584/m*</p></span></div>
```

```python
EMI=[]
dis_act_price=[]

for i,j in enumerate(data.find_all("div",attrs={'class':"styles_flexItem__2z__J"})):
    if i%2==0:
        EMI.append(j.text)
    else:
        dis_act_price.append(j.text)
```

```python
len(EMI),len(dis_act_price)
```

```python
cars=pd.DataFrame({'Cars':cnames,'Model':models,'Dist':dist,'Fuel':fuel,
                  'Transmission':transm,'Registration':regis,'EMI':EMI,
                  'Price':dis_act_price})
```

```python
cars
```

```python
cars.to_csv('cars24_1.csv',index=False)
```
