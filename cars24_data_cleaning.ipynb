```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("cars24_1.csv")
df.head()
```

```python
df.dtypes
```

```python
df['Manufac_Year']=df['Cars'].str[0:4]
df.head()
```

```python
df['Manufac_Year']=pd.to_numeric(df['Manufac_Year'])
```

```python
df['Cars']=df['Cars'].str[5::1]
df.head(1)
```

```python
df['Dist']=df['Dist'].str.replace(",","")
```

```python
df['Dist']=(df['Dist'].str[0:-3])
```

```python
df['Dist']=pd.to_numeric(df['Dist'])
```

```python
df.head(1)
```

```python
df.rename(columns={'Dist':'Distance_in_km'},inplace=True)
```

```python
df['Reg_unit']=df['Registration'].str[-2::1]
```

```python
df['Registration']=df['Registration'].str.replace("-","")
```

```python
df['Registration']=df['Registration'].str[0:2]
```

```python
df.head(1)
```

```python
df['EMI']=df['EMI'].str[5:-1:1]
```

```python
df['EMI']=df['EMI'].str.replace(",","")
df['EMI']=df['EMI'].str.replace("/","")
```

```python
df.head(1)
```

```python
df['EMI']=df['EMI'].str[:-1:1]
```

```python
df.rename(columns={'EMI':'EMI_Per_Month'},inplace=True)
```

```python
df['EMI_Per_Month']=df['EMI_Per_Month'].replace("vailable on E",np.nan)
```

```python
df
```

```python
df['EMI_Per_Month']=pd.to_numeric(df['EMI_Per_Month'])
```

```python
df.columns
```

```python
df['Price']=df['Price'].str.replace("lakh","L")
```

```python
df["Actual_Price_in_Lakhs"]=df["Price"].apply(lambda x: x.split("L")[0].replace("₹",""))
```

```python
df["Actual_Price_in_Lakhs"]=pd.to_numeric(df["Actual_Price_in_Lakhs"])
```

```python
df["Discounted_Price_in_Lakhs"]=(df["Price"].apply(lambda x: x.split("L")[1].replace("₹","")))
```

```python
df["Discounted_Price_in_Lakhs"]=df["Discounted_Price_in_Lakhs"].replace("+ other charges",np.nan)
```

```python
df["Discounted_Price_in_Lakhs"]=df["Discounted_Price_in_Lakhs"].fillna(df['Actual_Price_in_Lakhs'])
```

```python
df["Discounted_Price_in_Lakhs"]=pd.to_numeric(df["Discounted_Price_in_Lakhs"])
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.drop(columns=["Price"],inplace=True)
```

```python
df.head()
```

```python
df=df[["Cars","Model","Manufac_Year","Distance_in_km","Fuel","Transmission","Registration","Reg_unit","Actual_Price_in_Lakhs","Discounted_Price_in_Lakhs","EMI_Per_Month"]]
```

```python
df
```

```python
df.to_csv("Cars24_cleaned_data.csv",index=False)
```
